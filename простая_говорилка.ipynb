Пример подключения LLM API

```python
""" llm api """
import openai

YANDEX_CLOUD_FOLDER = "" # берём из консоли Яндекса
YANDEX_CLOUD_API_KEY = userdata.get('YANDEX_API') # безопасно извлекаем ключ
YANDEX_CLOUD_MODEL = "deepseek-v32/latest" # здесь можно выбрать другую модель

client = openai.OpenAI(
  api_key=YANDEX_CLOUD_API_KEY,
  base_url="https://ai.api.cloud.yandex.net/v1", # эндпоинт Яндекса
  project=YANDEX_CLOUD_FOLDER
)

def answer(message, client=client):
    """
    Пример функции для ответа LLM
    """
    response = client.responses.create(
        model=f"gpt://{YANDEX_CLOUD_FOLDER}/{YANDEX_CLOUD_MODEL}",
        temperature=0.3,
        instructions='здесь можно указать системную роль',
        input=message, # вот здесь будет output от whisper
        max_output_tokens=512
        )
    return response.output_text
```

Как это упаковать в бот

```python
""" tg bot """
import telebot
from openai import OpenAI
from google.colab import userdata

bot = telebot.TeleBot(userdata.get('TG_BOT'))

@bot.message_handler(func=lambda m: True)
def reply(message):
    try:
        answer_message = answer(message.text, client)
        bot.reply_to(message, answer_message)
    except Exception as e:
        bot.reply_to(message, f"Ошибка: {str(e)[:100]}")

bot.infinity_polling()
```
